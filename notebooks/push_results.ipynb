# Push results to GitHub

Run this **after** notebooks 01–04 have finished and been saved (Ctrl+S).
It commits `splits/`, `results/`, `weights/` and the executed notebooks from the Drive copy of the repo.

Needs a GitHub token stored in Colab's **Secrets** panel (key icon on the left) under the name `GITHUB_TOKEN`, with *Contents: read and write* access to this repo only. Never paste the token into a cell.

```
import os, subprocess
from google.colab import drive, userdata
drive.mount("/content/drive")
os.chdir("/content/drive/MyDrive/car-size")
token = userdata.get("GITHUB_TOKEN")

def run(cmd):
    out = subprocess.run(cmd, shell=True, capture_output=True, text=True)
    print((out.stdout + out.stderr).replace(token, "***"))

run('git config user.name "Mosaad Alrubaish" && git config user.email "YUPWE@users.noreply.github.com"')
run("git add splits results weights notebooks README.md && git status --short")
run('git commit -m "Results from our Colab runs"')
run(f"git pull --rebase https://x-access-token:{token}@github.com/YUPWE/car-size.git main")
run(f"git push https://x-access-token:{token}@github.com/YUPWE/car-size.git HEAD:main")
```
